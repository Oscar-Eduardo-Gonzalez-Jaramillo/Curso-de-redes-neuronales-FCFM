```python
df = pd.read_csv('/tf/global_house_purchase_dataset.csv')


categorical_cols = ['country', 'city', 'property_type', 'furnishing_status']
numerical_cols = ['property_size_sqft', 'price', 'previous_owners', 'constructed_year', 'rooms', 'bathrooms', 
                  'garage', 'crime_cases_reported', 'legal_cases_on_property', 'customer_salary', 
                  'loan_amount', 'loan_tenure_years', 'monthly_expenses', 'down_payment', 
                  'emi_to_income_ratio', 'satisfaction_score', 'neighbourhood_rating', 'connectivity_score']


df = df.dropna(subset=['decision'])


if df['decision'].dtype == 'object':
    df['decision'] = df['decision'].str.lower().str.strip()
    df = df[df['decision'].isin(['yes', 'no'])]
    y = df['decision'].map({'yes': 1, 'no': 0})
else:
    df = df[df['decision'].isin([0, 1])]
    y = df['decision'].astype(int)

X = df[categorical_cols + numerical_cols]

num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')), 
    ('scaler', StandardScaler())
])

cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')), 
    ('onehot', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_pipeline, numerical_cols),
        ('cat', cat_pipeline, categorical_cols)
    ])

X_processed = preprocessor.fit_transform(X)

feature_names = preprocessor.get_feature_names_out()
num_features = len(feature_names)

X_temp, X_test, y_temp, y_test = train_test_split(X_processed, y, test_size=0.2, random_state=42, stratify=y)

X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.125, random_state=42, stratify=y_temp)

y_train = y_train.to_numpy()
y_val = y_val.to_numpy()
y_test = y_test.to_numpy()
```

```python
def create_model(trial):  
    activation_options = ["relu", "leaky_relu", "elu"]

    inputs = layers.Input(shape=(num_features,))
    x = inputs
    residual_saved = None

    n_layers = trial.suggest_int("n_layers", 1, 9)

    for i in range(n_layers):
        n_units = trial.suggest_int(f"n_units_l{i}", 16, 300, log=True)
        batch_normalization = trial.suggest_categorical(f"batch_norm_l{i}", [0, 1])
        use_residual = trial.suggest_categorical(f"use_residual_l{i}", [0, 1])
        regularizer_l1 = trial.suggest_categorical(f"regularizer_L1_l{i}", [0, 1])
        regularizer_l2 = trial.suggest_categorical(f"regularizer_L2_l{i}", [0, 1])
        dropout = trial.suggest_categorical(f"regularizer_dropout_l{i}", [0, 1])

        residual = x
        
        regularizer = None
        if regularizer_l1 and regularizer_l2:
            lambda_l1 = trial.suggest_float(f"lambda_L1_l{i}", 0.0001, 0.9)
            lambda_l2 = trial.suggest_float(f"lambda_L2_l{i}", 0.0001, 1)
            regularizer = l1_l2(l1=lambda_l1, l2=lambda_l2)
            
        elif regularizer_l1:
            lambda_l1 = trial.suggest_float(f"lambda_L1_l{i}", 0.0001, 0.9)
            regularizer = l1(lambda_l1)
        elif regularizer_l2:
            lambda_l2 = trial.suggest_float(f"lambda_L2_l{i}", 0.0001, 1)  
            regularizer = l2(lambda_l2)
        elif regularizer_l1==0 and regularizer_l2==0 and dropout==0 :  
            regularizer = None
        
        
        x = layers.Dense(n_units, kernel_regularizer=regularizer)(x)
        
        if batch_normalization:
            x = layers.BatchNormalization()(x)
        
        activation_selected = trial.suggest_categorical(f"activation_l{i}", activation_options)
        if activation_selected == "relu":
            x = layers.ReLU()(x)
        elif activation_selected == "leaky_relu":
            alpha = trial.suggest_float(f"alpha_l{i}", 0.01, 0.3)
            x = layers.LeakyReLU(alpha=alpha)(x)
        elif activation_selected == "elu":
            alpha = trial.suggest_float(f"alpha_l{i}", 0.1, 1.0)
            x = layers.ELU(alpha=alpha)(x)

        if dropout:
            dropout_rate = trial.suggest_float(f"dropout_l{i}", 0.01, 0.4)
            x = layers.Dropout(dropout_rate)(x)
            
        if use_residual and residual is not None:
            if residual.shape[-1] != x.shape[-1]:
                residual = layers.Dense(x.shape[-1])(residual)  
            x = layers.Add()([x, residual])

         

    x = layers.Dense(1, activation='sigmoid')(x)

    model = keras.models.Model(inputs=inputs, outputs=x)

    lr = trial.suggest_float("lr", 1e-5, 1e-2, log=True)
    optimizer = Adam(learning_rate=lr)

    model.compile(optimizer=optimizer,
                  loss="binary_crossentropy",
                  metrics=["accuracy"])

    return model
```

```python
def objective(trial):
    with mlflow.start_run(nested=True, run_name=f"Prueba_{trial.number}"):
        run_id = mlflow.active_run().info.run_id
        trial.set_user_attr("run_id", run_id)
        batch_size = trial.suggest_int("batch_size", 32, 1000, log=True)
        
        model = create_model(trial)  # Assuming this function is defined elsewhere and uses trial for hyperparameters
        mlflow.log_params(trial.params)

        earlystop = EarlyStopping(
            monitor='val_accuracy',  # Monitor accuracy for early stopping
            mode='max',  # Maximize it
            restore_best_weights=True,
            patience=10,
            verbose=1
        )

        history = model.fit(
            X_train,
            y_train,
            epochs=40,
            batch_size=batch_size,
            validation_data=(X_test, y_test),  
            callbacks=[TFKerasPruningCallback(trial, "val_accuracy"), earlystop],  # Pruning also on val_acc
            verbose=0
        )
        
        # Save and log model as artifact
        model_path = f"model_trial_{trial.number}.keras"
        model.save(model_path)
        mlflow.log_artifact(model_path, "model")
        os.remove(model_path)  


        return history.history['val_accuracy'][-1]
```

```python
import os
import mlflow
import optuna
from tensorflow.keras.callbacks import EarlyStopping

mlflow.tensorflow.autolog(log_models=True)
mlflow.set_experiment("Tarea_5_arquitectura")

with mlflow.start_run(run_name="Tarea_5_Optuna_Prueba"):

    study = optuna.create_study(
        direction="minimize",
        study_name="Tarea_5_buena",
        pruner=optuna.pruners.MedianPruner(n_startup_trials=5, n_warmup_steps=5),
        storage="sqlite:///Tarea_5_optuna.db",
        load_if_exists=True
    )
    
    study.optimize(
        objective,
        n_trials=100,
        callbacks=[]  
    )
    
    best_trial = study.best_trial
    mlflow.log_params(best_trial.params)
    mlflow.log_metric("best_test_loss", best_trial.value)

    
    with mlflow.start_run(nested=True, run_name="Best_Model_Training"):
        best_model = create_model(best_trial)
        batch_size = best_trial.params["batch_size"]
        mlflow.log_params(best_trial.params)  

        earlystop = EarlyStopping(
            monitor='val_loss',
            mode='min',
            restore_best_weights=True,
            patience=30,  
            verbose=1
        )
        
        best_model.fit(
            X_train,
            y_train,
            epochs=300,  # Más epochs para entrenar bien el mejor modelo
            batch_size=batch_size,
            validation_data=(X_test, y_test),
            callbacks=[earlystop],  # mlflow.tensorflow.autolog() hace el logging
            verbose=1
        )

        # Guardar y loguear el modelo final
        best_model_path = "best_model.keras"
        best_model.save(best_model_path)
        mlflow.log_artifact(best_model_path, "best_model")
        os.remove(best_model_path)

    print(f"Mejor test_loss: {best_trial.value}")
    print(f"Mejores params: {best_trial.params}")
```
