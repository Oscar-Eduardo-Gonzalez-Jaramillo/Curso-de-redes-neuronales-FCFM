```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout,Activation
from tensorflow.keras.optimizers import RMSprop, Adam
from matplotlib import pyplot as plt
import numpy as np
from tensorflow.keras.saving import register_keras_serializable
```

```python
loss_tracker = tf.keras.metrics.Mean(name="loss")
@register_keras_serializable()
class Problema_3_grad3(tf.keras.Model):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        
        g = tf.random.Generator.from_non_deterministic_state()
        self.coeficientes = self.add_weight(name="coeficientes",
            shape=(4, 1),
            initializer=tf.keras.initializers.RandomNormal(), 
            trainable=True,)
        
    def call(self, inputs):
        inputs = tf.convert_to_tensor(inputs, dtype=tf.float32)
        if inputs.shape == ():
            inputs = tf.reshape(inputs, [1, 1])
        elif len(inputs.shape) == 1:
            inputs = tf.expand_dims(inputs, axis=1)
        batch = tf.shape(inputs)[0]
        columna1 = tf.ones([batch,1])
        columna3 = tf.multiply(inputs, inputs)
        columna4 = tf.multiply(columna3, inputs)

        A = tf.concat([columna1, inputs, columna3, columna4], axis=1)
        self.polinomio = tf.tensordot(A, self.coeficientes, 1)
        return self.polinomio

    @property
    def metrics(self):
        return [loss_tracker] 

    def train_step(self, data):
        batch_size = 1  
        x = tf.random.uniform((batch_size, 1), minval=-1, maxval=1) 
        eq = tf.math.cos(2 * x)

        with tf.GradientTape() as tape:
            y_pred = self(x, training=True)
            funcionloss = tf.keras.losses.MeanSquaredError()
            loss=funcionloss(y_pred,eq)

        grads = tape.gradient(loss, [self.coeficientes])
        self.optimizer.apply_gradients(zip(grads, [self.coeficientes]))

        
        

        loss_tracker.update_state(loss)

        return {"loss": loss_tracker.result()}
```

```python
model_3 = Problema_3_grad3()  
model_3.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001))
model_3.summary()
```

```python
x_data = np.random.uniform(-1, 1, 1000).astype(np.float32) #Datos que no usa el modelo pero necesarios para la inicialización 
y_data = np.cos(2 * x_data).astype(np.float32)
dataset=[x_data, y_data]
history = model_3.fit(dataset, epochs=1000, steps_per_epoch=200, verbose=1)
```
